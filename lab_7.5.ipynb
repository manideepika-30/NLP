```python
import numpy as np
import pandas as pd
import re

import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

from nltk.corpus import wordnet as wn
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
documents = [
    "Machine learning is a branch of artificial intelligence",
    "Artificial intelligence enables machines to think",
    "Deep learning is a subset of machine learning",
    "Natural language processing deals with text data",
    "Text similarity measures how similar documents are",
    "Doctors diagnose diseases and prescribe medicine",
    "Physicians treat patients in hospitals",
    "Cats are domestic animals",
    "Dogs are loyal pets",
    "Big data analytics helps businesses",
    "Data science uses statistics and programming",
    "Students study hard for exams",
    "Learners prepare for tests",
    "Cars require fuel to run",
    "Automobiles need gasoline",
    "Python is a popular programming language",
    "Java is also a programming language",
    "Football is a popular sport",
    "Soccer is loved worldwide",
    "The sun rises in the east",
    "The moon orbits the earth",
    "Teachers educate students",
    "Professors teach learners",
    "Hospitals treat sick people",
    "Clinics provide medical care"
]

df = pd.DataFrame(documents, columns=["Text"])
df.head()
```

```python
nltk.download('punkt_tab')

stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)
    tokens = word_tokenize(text)
    tokens = [lemmatizer.lemmatize(word) for word in tokens if word not in stop_words]
    return " ".join(tokens)

df["Clean_Text"] = df["Text"].apply(preprocess)
df.head()
```

```python
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(df["Clean_Text"])

tfidf_matrix.shape
```

```python
cosine_sim = cosine_similarity(tfidf_matrix)

cosine_df = pd.DataFrame(cosine_sim, index=df["Text"], columns=df["Text"])
cosine_df.head()
```

```python
def jaccard_similarity(text1, text2):
    set1 = set(text1.split())
    set2 = set(text2.split())
    return len(set1.intersection(set2)) / len(set1.union(set2))
```

```python
pairs = [
    (0, 2),
    (5, 6),
    (13, 14),
    (11, 12),
    (21, 22)
]

for i, j in pairs:
    score = jaccard_similarity(df["Clean_Text"][i], df["Clean_Text"][j])
    print(f"Jaccard Similarity between:\n'{df['Text'][i]}'\nAND\n'{df['Text'][j]}' = {score:.2f}\n")
```

```python
def wordnet_similarity(word1, word2):
    syn1 = wn.synsets(word1)
    syn2 = wn.synsets(word2)
    if syn1 and syn2:
        return syn1[0].wup_similarity(syn2[0])
    return None
```

```python
word_pairs = [
    ("doctor", "physician"),
    ("car", "automobile"),
    ("student", "learner"),
    ("teacher", "professor"),
    ("dog", "cat")
]

for w1, w2 in word_pairs:
    print(f"WordNet similarity between '{w1}' and '{w2}':",
          wordnet_similarity(w1, w2))
```

### Conclusion
This lab demonstrated three text similarity techniques. Cosine similarity
performed well with TF-IDF vectors. Jaccard similarity highlighted exact
word overlap limitations. WordNet-based similarity successfully captured
semantic relationships. The choice of similarity method depends on the
application and text type.
